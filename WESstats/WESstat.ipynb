# Q>20 & Q>30

```python
# Расчет Q>20 и Q>30 из FFQ и LFQ
samtools stats /data/WES_260526/align/sample_trimmed_sort.bam 2>/dev/null | grep -E "^(FFQ|LFQ)" | awk '
{
    for(i=2; i<=NF; i++) {
        qual = i-2
        count = $i
        if(qual > 20) q20 += count
        if(qual > 30) q30 += count
        total += count
    }
}
END {
    if(total > 0) {
        printf "Q>20: %.2f%%\n", q20/total*100
        printf "Q>30: %.2f%%\n", q30/total*100
        printf "Total bases: %d\n", total
    } else {
        print "Q>20: N/A"
        print "Q>30: N/A"
    }
}'
```
